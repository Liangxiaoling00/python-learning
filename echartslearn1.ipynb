```python
from IPython.display import HTML
```

```python
from pyecharts import options as opts
```

```python
from pyecharts.charts import Bar
```

```python
#最最初始版
bar=(
    Bar()
    .add_xaxis(['a','b','c','d','e'])
    .add_yaxis('way1',[5,40,20,35,60])
)
HTML(bar.render_embed())
```

```python
bar=(
    Bar()
    .add_xaxis(['a','b','c','d','e'])
    .add_yaxis('way1',[5,40,20,35,60])

    #配置项
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title='主标题',
            subtitle='副标题'
        )
    )
    #或者
    #.set_global_opts(
    #title_opts={
        #'text':'主标题',
        #'subtext':'副标题'
    #}
    #)
)
bar.render('chartslearn1')
HTML(bar.render_embed())
```

```python
from pyecharts.globals import ThemeType
```

```python
bar=(
    Bar(
        init_opts=opts.InitOpts(
            theme=ThemeType.DARK
        )
    )
    .add_xaxis(['a','b','c','d','e'])
    .add_yaxis('way1',[5,40,20,35,60])
)
HTML(bar.render_embed())
```

```python
bar=(
    Bar()
    .add_xaxis(['a','b','c','d','e'])
    .add_yaxis('way1',[5,40,20,35,60])
    .add_yaxis('way2',[34,6,32,12,50])
)
HTML(bar.render_embed())
```
