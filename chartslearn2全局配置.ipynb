```python
from pyecharts.charts import Bar,Line
from pyecharts import options as opts

from pyecharts.faker import Faker
from pyecharts.globals import ThemeType,RenderType
```

```python
Faker.choose()
```

```python
c=(
    Bar(
        #初始化配置项
        init_opts=opts.InitOpts(
            width='700px',
            height='400px', #图表大小
            renderer=RenderType.CANVAS, #渲染风格，可选canvas svg
            page_title='一个网页标题',
            theme=ThemeType.WALDEN,#主题
            bg_color='pink'#背景颜色
        )
            
    )
    .add_xaxis(Faker.choose())
    .add_yaxis('商家A',Faker.values())
    .add_yaxis('商家B',Faker.values())

    #全局配置项
    .set_global_opts(

        #标题配置
        title_opts=opts.TitleOpts(
            title='柱形图', #主标题
            title_link='https://www.baidu.com', #主标题点击跳转链接
            title_target='blank', #blank表示新窗口打开，self当前窗口打开
            #sub标题同理

            #位置
            pos_left='20px',
            pos_top='10px',
            #pos_right
            #pos_bottom
            padding=20, #内边距,标题到顶部的距离
            item_gap=10,#主标题副标题的间隙
        ),
        #DataZoomOpts：区域缩放配置项
        datazoom_opts=opts.DataZoomOpts(
            is_show=True,#是否显示
            type_='slider',#组件类型：slider,inside（鼠标滚动）
            is_realtime=True,#拖动的时候是否实时更新图表数据，数据太多可能会卡顿
            range_start=40,#数据窗口的开始百分比
            range_end=80,
            orient='horizontal',#或vertical垂直
            is_zoom_lock=False,#滑动条是否锁定范围大小
        ),
        
        #图例配置项
        legend_opts=opts.LegendOpts(
            #图例类型：plain普通图例，scoll可以滚动的图例，用于图例较多的情况
            type_='plain',
            is_show=True,
            pos_top='5%',#图例位置：pos_left,pos_right,pos_top,pos_bottom
            orient='horizontal',#或vertical,图例摆列方式
            #选择模式
            #true：开启图例点击
            #False：关闭
            #single：单选
            #multiple：多选
            selected_mode='multiple',
            #图标和文字的位置
            align='right',
            padding=20,#内边距
            item_gap=5,#图例中每一项的间距
            item_width=30,#项的宽度
            item_height=5,#项的高度
            inactive_color='#ccc',#图例的颜色

            #PyEcharts常见的图标：circle,rect,roundRect,triangle,diamond,arrow
            legend_icon='diamond',
        ),
        
        #VisualMapOpts：视觉映射配置项
        visualmap_opts=opts.VisualMapOpts(
            is_show=True,
            type_='color',#color或size
            min_=0,
            max_=150,
            range_opacity=0.7,#图元和文字透明度
            range_text=['max','min'],#两端的文本
            range_color=['yellow','red','black'],#过度颜色

            orient='horizontal',#或vertical
            pos_right='5%',
            pos_top='0%',
            is_piecewise=True,#是否为分段型
            is_inverse=True,#是否反转
        ),

        #TooltipOpts:提示框配置
        tooltip_opts=opts.TooltipOpts(
            is_show=True,

            #触发类型
            #item：数据项，用于散点图，柱形图，饼图
            #axis：坐标轴，提示线，用于条形图，折线图等
            trigger='item',
            #触发条件：
            #mousemove，click，mousemove/click
            trigger_on='click',

            is_show_content=True,#是否显示提示框浮层

            #标签内容的格式：
            #字符串中的模板变量：
            #{a}:系列名series_name
            #{b}:数据名
            #{c}:值
            formatter='{a}:{b}-{c}',

            background_color='black',#背景颜色
            border_color='white',#边框颜色
            border_width=1,#边框宽度
        ),

        #AxisOpts:坐标轴配置项
        xaxis_opts=opts.AxisOpts(
            is_show=True,#是否显示x轴

            #坐标轴类型：
            # value：数值轴，适用于连续数据
            # category：类目轴，适用于离散数据
            # time：时间轴，适用于连续的时序数据
            type_='category'
        ),
        yaxis_opts=opts.AxisOpts(
            is_show=True,
            axisline_opts=opts.AxisLineOpts(
                is_show=False #不显示y轴的线
            ),
            axistick_opts=opts.AxisTickOpts(
                is_show=True #显示y轴的刻度
            )
        )
    )
)
HTML(c.render_embed())
```
